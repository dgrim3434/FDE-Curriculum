```python
from data.constants import SCIFACT_PATH
from sentence_transformers import SentenceTransformer
import json


tok = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2").tokenizer

path = SCIFACT_PATH + "corpus.jsonl"

result = []

with open(path, 'r', encoding='utf-8') as f:
    
    for line in f:
        
        row = json.loads(line)

        id = str(row['_id'])
        text = row['title'] + " " + row['text'].strip()
            
        result.append((id, text))      
```
